```python
# ---------------------------
# Imports
# ---------------------------
import warnings
warnings.filterwarnings('ignore')

import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from xgboost import XGBRegressor
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

# ---------------------------
# Caminhos dos arquivos
# ---------------------------
caminho_sp_capital = "/content/BANCO_SPCAPITAL.xlsx"
caminho_sp_exceto  = "/content/BANCO_SP_EXCETOCAPITAL.xlsx"
caminho_sp_capital_setor  = "/content/BANCO_SPCAPITAL_setorcensitario.xlsx"
caminho_sp_exceto_setor  = "/content/BANCO_SP_EXCETOCAPITAL_setorcensitario.xlsx"

# ---------------------------
# Colunas-resposta
# ---------------------------
colunas_resposta = {
    'domicilios_precarios': 'DOMICILIOS PRECARIOS',
    'coabitacao': 'COABITACAO',
    'onus_excessivo': 'ONUS EXCESSIVO',
    'adensamento': 'ADENSAMENTO',
    'deficit_total': 'DEFICIT TOTAL'
}

# ---------------------------
# Funções auxiliares
# ---------------------------
def preparar_dataset(df, coluna_resposta, coluna_id=None, colunas_remover=None):
    """
    Prepara o dataset para treino:
    - Remove coluna ID e outras colunas opcionais
    - Converte strings numéricas para números
    - Separa X (features) e y (resposta)
    - Remove valores NA da resposta e preenche medianas nas features
    """
    df2 = df.copy()

    # Remove coluna ID
    if coluna_id and coluna_id in df2.columns:
        df2 = df2.drop(columns=[coluna_id])

    # Remove colunas opcionais
    if colunas_remover:
        for c in colunas_remover:
            if c in df2.columns:
                df2 = df2.drop(columns=[c])

    # Conversão de strings numéricas
    for c in df2.columns:
        if df2[c].dtype == object:
            df2[c] = df2[c].astype(str).str.replace(',', '.').str.replace(' ', '')
            df2[c] = pd.to_numeric(df2[c], errors='ignore')

    if coluna_resposta not in df2.columns:
        raise ValueError(f"Coluna resposta {coluna_resposta} não encontrada no DataFrame")

    y = pd.to_numeric(df2[coluna_resposta], errors='coerce')
    X = df2.drop(columns=[coluna_resposta])

    X_numerico = X.select_dtypes(include=[np.number]).copy()

    # Mantém apenas linhas onde y não é nulo
    mascara = y.notna()
    X_numerico = X_numerico.loc[mascara]
    y = y.loc[mascara]

    # Preenche valores nulos nas features com mediana
    X_numerico = X_numerico.fillna(X_numerico.median())

    return X_numerico, y, X_numerico.columns.tolist()


def treinar_avaliar_xgb(X, y, serie_id=None, nome_resposta='', random_state=42):
    """
    Treina um modelo XGBoost e avalia desempenho:
    - Divide em treino/teste (80/20)
    - Calcula RMSE, MAE, R²
    - Plota Observado vs Predito
    - Mostra top 10 features
    - Exibe erros por ID se fornecido
    """
    # Divisão treino/teste
    X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.2, random_state=random_state)

    # Criação e treinamento do modelo
    modelo = XGBRegressor(
        objective='reg:squarederror',
        n_estimators=500,
        learning_rate=0.05,
        max_depth=6,
        random_state=random_state
    )
    modelo.fit(X_treino, y_treino)

    # Predição no conjunto de teste
    y_pred = modelo.predict(X_teste)

    # Cálculo de métricas
    rmse = mean_squared_error(y_teste, y_pred) ** 0.5
    mae = mean_absolute_error(y_teste, y_pred)
    r2 = r2_score(y_teste, y_pred)

    print(f"{nome_resposta} -> Teste: RMSE={rmse:.2f}, MAE={mae:.2f}, R2={r2:.3f}")

    # Importância das features
    importancia_features = pd.Series(modelo.feature_importances_, index=X.columns).sort_values(ascending=False)
    print("\nTop 10 Features:\n", importancia_features.head(10))

    # Gráfico Observado vs Predito
    plt.figure(figsize=(6,6))
    sns.scatterplot(x=y_teste, y=y_pred, alpha=0.6)
    plt.plot([y_teste.min(), y_teste.max()], [y_teste.min(), y_teste.max()], 'r--')
    plt.xlabel('Observado')
    plt.ylabel('Predito')
    plt.title(f'{nome_resposta} - Observado vs Predito (Teste)')
    plt.show()

    # DataFrame de erros por ID
    if serie_id is not None:
        df_erros = pd.DataFrame({
            'ID': serie_id.loc[y_teste.index],
            'observado': y_teste,
            'predito': y_pred,
            'erro': y_teste - y_pred
        })
        display(df_erros.sort_values('erro', key=abs, ascending=False).head(10))
    else:
        df_erros = None

    resultados = {
        'modelo': modelo,
        'X_teste': X_teste,
        'y_teste': y_teste,
        'y_pred': y_pred,
        'rmse': rmse,
        'mae': mae,
        'r2': r2,
        'importancia_features': importancia_features,
        'df_erros': df_erros
    }
    return resultados

# ---------------------------
# Etapa 1: Treinamento
# ---------------------------
def etapa1_treinar_por_arquivo(caminho_arquivo, rotulo, colunas_resposta):

    print(f"\n--- Carregando arquivo: {rotulo} ---")
    df = pd.read_excel(caminho_arquivo, engine='openpyxl')
    print(f"Dimensões: {df.shape}")
    print("Exemplo de colunas (primeiras 40):\n", df.columns.tolist()[:40])

    coluna_id = df.columns[0]
    print('Coluna ID assumida como:', coluna_id)

    resultados_modelos = {}

    for variavel, nome_coluna in colunas_resposta.items():
        print(f"\nTreinando para variável-resposta: {variavel} -> coluna '{nome_coluna}'")
        try:
            X, y, nomes_features = preparar_dataset(df, coluna_resposta=nome_coluna, coluna_id=coluna_id)
            res = treinar_avaliar_xgb(X, y, serie_id=df[coluna_id], nome_resposta=variavel)
            resultados_modelos[variavel] = {
                'coluna_resposta': nome_coluna,
                'coluna_id': coluna_id,
                'colunas_features': nomes_features,
                'resultados': res
            }
            print(f"Resultados {variavel}: RMSE={res['rmse']:.3f}, MAE={res['mae']:.3f}, R2={res['r2']:.3f}")
        except Exception as e:
            print(f"Erro treinando {variavel}: {e}")

    return resultados_modelos, df


# ---------------------------
# Executando Etapa 1
# ---------------------------
modelos_capital, df_capital = etapa1_treinar_por_arquivo(caminho_sp_capital, 'SP Capital (AP)', colunas_resposta)
modelos_exceto, df_exceto = etapa1_treinar_por_arquivo(caminho_sp_exceto, 'SP Exceto Capital (AP)', colunas_resposta)

```

```python
print("\n\n==========================")
print("COMPARAÇÃO: CAPITAL vs EXCETO CAPITAL")
print("==========================\n")

comparacao = []

for variavel in colunas_resposta.keys():

    res_cap = modelos_capital[variavel]['resultados']
    res_exc = modelos_exceto[variavel]['resultados']

    comparacao.append({
        'variavel': variavel,
        'RMSE_capital': res_cap['rmse'],
        'RMSE_exceto': res_exc['rmse'],
        'MAE_capital': res_cap['mae'],
        'MAE_exceto': res_exc['mae'],
        'R2_capital': res_cap['r2'],
        'R2_exceto': res_exc['r2'],
    })

df_comparacao = pd.DataFrame(comparacao)
display(df_comparacao)

print("\nResumo rápido:")
for linha in comparacao:
    print(f"- {linha['variavel']}: Capital R2={linha['R2_capital']:.3f} | Interior R2={linha['R2_exceto']:.3f}")
```
